```python
# Correct way to fill missing values
matches = matches.assign(
    city=matches['city'].fillna('Unknown'),
    player_of_match=matches['player_of_match'].fillna('No Award'),
    winner=matches['winner'].fillna('No Result'),
    result_margin=matches['result_margin'].fillna(0),
    target_runs=matches['target_runs'].bfill(),  # Using .bfill() instead of 'method=bfill'
    target_overs=matches['target_overs'].bfill(),
    method=matches['method'].fillna('Not Applied')
)

# Verify missing values are handled
print(matches.isnull().sum())




```

```python
deliveries = deliveries.assign(
    extras_type=deliveries['extras_type'].fillna('No Extras'),
    player_dismissed=deliveries['player_dismissed'].fillna('Not Out'),
    dismissal_kind=deliveries['dismissal_kind'].fillna('Not Applicable'),
    fielder=deliveries['fielder'].fillna('No Fielder')
)

# Verify that missing values are handled
print(deliveries.isnull().sum())
```

# Win Percentage by Team (Pie Chart)

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load matches data
matches = pd.read_csv('matches.csv')

# Calculate total matches and wins per team
total_matches = matches['team1'].value_counts() + matches['team2'].value_counts()
wins = matches['winner'].value_counts()

# Calculate win percentage
win_percentage = (wins / total_matches) * 100
win_percentage.dropna(inplace=True)

# Pie chart for team win percentage
plt.figure(figsize=(10, 8))
plt.pie(win_percentage, labels=win_percentage.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.tab20.colors)
plt.title('Win Percentage of IPL Teams')
plt.show()
```

# Team Performance Over the Years (Line Chart)

```python
# Wins per year for each team
team_wins_per_year = matches.groupby(['season', 'winner']).size().reset_index(name='wins')

# Pivot table for line plot
team_wins_pivot = team_wins_per_year.pivot(index='season', columns='winner', values='wins').fillna(0)

# Line plot
plt.figure(figsize=(12, 6))
team_wins_pivot.plot(marker='o', colormap='tab20', linewidth=2)
plt.title('Team Performance Over the Years')
plt.xlabel('Season')
plt.ylabel('Number of Wins')
plt.legend(title='Teams', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.show()
```

# Team Performance Before & After Adding/Transferring a Key Player



```python
import pandas as pd
import matplotlib.pyplot as plt

# Ensure 'season' is numeric
matches['season'] = pd.to_numeric(matches['season'], errors='coerce')

# Example: Analyzing how team performance changed after adding 'Chris Gayle'
team = 'Royal Challengers Bangalore'
gayle_debut_season = 2011

# Filter matches where the team won before and after player joined
season_with_player = matches[(matches['winner'] == team) & (matches['season'] >= gayle_debut_season)]
season_before_player = matches[(matches['winner'] == team) & (matches['season'] < gayle_debut_season)]

# Wins per season before and after
wins_before = season_before_player.groupby('season')['winner'].count()
wins_after = season_with_player.groupby('season')['winner'].count()

# Plotting
plt.figure(figsize=(10,6))
plt.plot(wins_before.index, wins_before.values, marker='o', label='Before Chris Gayle')
plt.plot(wins_after.index, wins_after.values, marker='x', label='After Chris Gayle')

plt.xlabel('Season')
plt.ylabel('Number of Wins')
plt.title(f'{team} Performance Before and After Chris Gayle')
plt.legend()
plt.grid(True)
```

# Venue Performance 

```python

import seaborn as sns

# Wins per venue
venue_wins = matches.groupby('venue')['winner'].value_counts().unstack().fillna(0)

# Total matches per venue
venue_matches = matches['venue'].value_counts()

# Win percentage at each venue
venue_win_percentage = (venue_wins.T / venue_matches).T * 100

# Bar chart of total matches played at each venue
plt.figure(figsize=(12, 6))
venue_matches.sort_values(ascending=False).plot(kind='bar', color='skyblue')
plt.title('Total Matches Played at Each Venue')
plt.xlabel('Venue')
plt.ylabel('Number of Matches')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# Heatmap of win percentage by venue
plt.figure(figsize=(12, 8))
sns.heatmap(venue_win_percentage, cmap='viridis', annot=True, fmt='.1f', linewidths=.5)
plt.title('Win Percentage of Teams Across Different Venues')
plt.xlabel('Team')
plt.ylabel('Venue')
plt.tight_layout()
plt.show()
```

# Run Rate & Scoring Analysis

```python
ipl_data = deliveries.merge(matches[['id', 'season']], left_on='match_id', right_on='id')

# Calculate total runs and balls faced per team per season
team_runs = ipl_data.groupby(['season', 'batting_team'])['total_runs'].sum().reset_index()
team_balls = ipl_data.groupby(['season', 'batting_team'])['ball'].count().reset_index()

# Merge runs and balls to calculate run rate
team_run_rate = team_runs.merge(team_balls, on=['season', 'batting_team'])
team_run_rate['run_rate'] = (team_run_rate['total_runs'] / team_run_rate['ball']) * 6

# Plot run rate trends for top teams
plt.figure(figsize=(12, 6))
for team in team_run_rate['batting_team'].unique():
    team_data = team_run_rate[team_run_rate['batting_team'] == team]
    plt.plot(team_data['season'], team_data['run_rate'], marker='o', label=team)

plt.xlabel("Season")
plt.ylabel("Run Rate")
plt.title("Team Run Rate Across Seasons")
plt.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.tight_layout()
plt.show()


```

# Best Batting Partnerships

```python
deliveries['partnership'] = deliveries.apply(lambda row: tuple(sorted([row['batter'], row['non_striker']])), axis=1)

# Calculate total runs for each partnership
partnership_runs = deliveries.groupby('partnership')['total_runs'].sum().reset_index()
partnership_runs.columns = ['partnership', 'total_runs']

# Calculate number of boundaries (4s and 6s) for each partnership
boundaries = deliveries[deliveries['batsman_runs'].isin([4, 6])]
partnership_boundaries = boundaries.groupby('partnership')['batsman_runs'].count().reset_index()
partnership_boundaries.columns = ['partnership', 'boundary_count']

# Merge the two datasets
partnership_stats = pd.merge(partnership_runs, partnership_boundaries, on='partnership', how='left').fillna(0)

# Top 10 partnerships by total runs
top_partnerships = partnership_stats.sort_values(by='total_runs', ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_partnerships['partnership'].astype(str), y=top_partnerships['total_runs'], palette='viridis')
plt.xticks(rotation=45)
plt.xlabel('Partnership')
plt.ylabel('Total Runs')
plt.title('Top 10 Batting Partnerships by Total Runs')
plt.show()

# Top 10 partnerships by boundary count
top_boundaries = partnership_stats.sort_values(by='boundary_count', ascending=False).head(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_boundaries['partnership'].astype(str), y=top_boundaries['boundary_count'], palette='magma')
plt.xticks(rotation=45)
plt.xlabel('Partnership')
plt.ylabel('Boundary Count')
plt.title('Top 10 Batting Partnerships by Boundaries')
plt.show()
```
